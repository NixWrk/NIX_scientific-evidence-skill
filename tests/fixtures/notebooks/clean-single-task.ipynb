# Оценка среднего времени обработки

Задача отчёта — определить среднее время обработки для синтетических результатов стендового измерения трёх последовательных операций. Измеряемая величина — длительность операции в миллисекундах; место и дата для учебной фикстуры не задаются. В расчёт входит один набор наблюдений, а межприборная устойчивость не проверяется. Работа завершена после расчёта среднего и построения рисунка.

## Измеряемая величина и расчёт среднего

Объектом расчёта служит длительность отдельной операции. Среднее арифметическое определяется как сумма длительностей, делённая на число наблюдений; все значения выражены в миллисекундах. Записи считаются независимыми, исключения не выполняются. Выход расчёта используется для описания только данной серии.

Для оценки возможного порядка измерений необходимо рассмотреть зависимость длительности операции от её номера. Такое сопоставление позволяет проверить, не скрывает ли среднее значение последовательное изменение времени обработки.

```python
values_ms = [1.0, 2.0, 3.0]
mean_ms = sum(values_ms) / len(values_ms)
```

Рисунок 1. Зависимость длительности операции от номера наблюдения в серии DATA-001.

```python
from IPython.display import Markdown, display
display(Markdown(f"По данным на рисунке 1 длительность последовательно возрастает, а среднее время составляет {mean_ms:.1f} мс. Результат описывает только DATA-001; межприборная вариабельность не установлена."))
```

```python
display(Markdown(f"""# Итог
Установлено среднее время {mean_ms:.1f} мс для DATA-001. Не установлено поведение на других приборах."""))
```
