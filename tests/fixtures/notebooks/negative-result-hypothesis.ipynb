# Проверка различия между двумя группами

Задача отчёта — проверить различие по результатам синтетического сравнительного эксперимента с двумя группами наблюдений. Причинные связи и перенос результата за пределы этой фикстуры не оцениваются. Работа завершена после расчёта заданного критерия сравнения.

## Модель сравнения и критерий вывода

Нулевая гипотеза описывает отсутствие различия между группами. Применяется заданный статистический критерий; наблюдения предполагаются независимыми. Значение p безразмерно и используется только для решения по принятому порогу, а не как мера величины эффекта.

```
p_value = 0.31
print(f'p = {p_value}')
```

```
from IPython.display import Markdown, display
display(Markdown(f"""**Наблюдение.** p = {p_value:.2f} при заданном критерии.
**Интерпретация.** Различие в DATA-003 не установлено.
**Ограничение.** Отсутствие различия не доказано.
**Гипотеза автора текущей тетради.** Мощность могла быть недостаточной; требуется отдельная проверка."""))
```

```
display(Markdown(f"""# Итог
Различие не установлено при p = {p_value:.2f}. Не установлена причина отрицательного результата; гипотеза не включена в вывод."""))
```
